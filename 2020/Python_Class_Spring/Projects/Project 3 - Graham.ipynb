```python
#Problem 1
# 1 Generate a DataFrame which has 4 columns with length 500, and x=pd.date_range('9/1/2018', periods=500)). Name the columns as A, B, C, D
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.DataFrame(np.random.randn(500, 4), index=pd.date_range('9/1/2018', periods=500), columns =['A','B','C','D'])
df
```

```python
#2 Name the cumulative sum of your data frame as a new data frame
df = df.cumsum()
df
```

```python
#3 Plot your new data frame, and save your figure as ‘Project3_your name_Fig1.png’
df.plot()
plt.savefig('Project3_Sean Graham_Fig1.png')
```

```python
#4 Plot only columns A and C, and save your figure as ‘Project3_your name_Fig2.png’
df.plot(y=['A','C'])
plt.savefig('Project3_Sean Graham_Fig2.png')
```

```python
#Problem 2
#1 Load the data Top 100 Retailers 2015.csv
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
Top_100 = pd.read_csv('Top 100 Retailers 2015.csv')
Top_100
```

```python
#2 Select the companies with 10<Rank <=20 and name the 10 companies as a new dataframe called Top_20, and sort Top_20 by Rank with ascending =True
Top_20=Top_100[(Top_100['Rank']>10) & (Top_100['Rank']<=20)]
Top_20.sort_values('Rank', ascending=True).head(10)
```

```python
#3 For the dataframe Top_20, plot the bar chart of both USA_Retail_Sales_million and Worldwide_ Retail_ Sales_million, with x='Company'
Top_20.index = Top_20['Company']
Top_20.plot.bar(y=['USA_Retail_Sales_million','Worldwide_ Retail_ Sales_million'], rot=70)
```

```python
#4 For the dataframe Top_20, plot the bar chart of Rank, with x='Company'
Top_20.index = Top_20['Company']
Top_20.plot.bar(y=['Rank'], rot=70)
```

```python
#5 For the dataframe Top_20, plot the bar chart of Worldwide_ Retail_ Sales_million, with x='Company'
Top_20.index = Top_20['Company']
Top_20.plot.bar(y=['Worldwide_ Retail_ Sales_million'], rot=70)
```

```python
#6 For the dataframe Top_20, plot the horizontal bar chart of both USA_Retail_Sales_million and Worldwide_ Retail_ Sales_million, with x='Company', and save the figure as ‘Project3_your name_Fig3.png’.
Top_20.index = Top_20['Company']
Top_20.plot.barh(y=['USA_Retail_Sales_million','Worldwide_ Retail_ Sales_million'])
plt.savefig('Project3_Sean Graham_Fig3.png')
```

```python
#7 For the dataframe Top_20, plot the horizontal bar chart of Worldwide_ Retail_ Sales_million, with the index rotation angle=45, with x='Company'
Top_20.index = Top_20['Company']
Top_20.plot.barh(y=['Worldwide_ Retail_ Sales_million'], rot=45)
```

```python
#8 For the dataframe Top_20, plot the horizontal bar chart of both USA_Retail_Sales_million and Worldwide_ Retail_ Sales_million, with x='Company', rot=45 and stacked=True
Top_20.index = Top_20['Company']
Top_20.plot.barh(y=['USA_Retail_Sales_million','Worldwide_ Retail_ Sales_million'], rot=45, stacked=True)
```

```python
#9 For the dataframe Top_20, plot the stacked area of both USA_Retail_Sales_million and Worldwide_ Retail_ Sales_million, with x='Company', rot=45 and stacked=True, and save the figure as ‘Project3_your name_Fig4.png’.
Top_20.index = Top_20['Company']
Top_20.plot.area(y=['USA_Retail_Sales_million','Worldwide_ Retail_ Sales_million'], rot=45, stacked=True)
plt.savefig('Project3_Sean Graham_Fig4.png')
```

```python
#10 Name the index of the Top_20 as 'Company' (Hint: Top_20.index=Top_20['Company'] ), For the dataframe Top_20, plot pie chart for both USA_Retail_Sales_million and Worldwide_ Retail_ Sales_million, with figsize=(10, 10), autopct='%1.2f%%', legend=False, subplots=True. Save the figure as ‘Project3_your name_Fig5.png’.
Top_20.index=Top_20['Company']
Top_20.plot.pie(y=['USA_Retail_Sales_million','Worldwide_ Retail_ Sales_million'], figsize=(10, 10), autopct='%1.2f%%', legend=False, subplots=True)
plt.savefig('Project3_Sean Graham_Fig5.png')
```

```python
#11 For the dataframe Top_20, scatter plot the 'USA_Retail_Sales_million'and 'Worldwide_ Retail_Sales_million' along Rank, and put the two graphs in one figure. Save the figure as ‘Project3_your name_Fig6.png’.
plot1=Top_20.plot.scatter(x='Rank', y='USA_Retail_Sales_million', c='r', marker='o', label='USA Retail Sales')
Top_20.plot.scatter(x='Rank', y='Worldwide_ Retail_ Sales_million', c='b', marker='x', ax=plot1, label='Worldwide Retail Sales')
plt.savefig('Project3_Sean Graham_Fig6.png')
```
