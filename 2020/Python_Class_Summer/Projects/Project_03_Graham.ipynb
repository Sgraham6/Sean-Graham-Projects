## Part 3: Working with Categorical Variables
This section involves creating lists to specify the order for each of the three categorical variables.

```python
clarity_levels = ['I1','SI2','SI1','VS2','VS1','VVS2','VVS1','IF']
cut_levels = ['Fair','Good','Very Good','Premium','Ideal']
color_levels = ['J','I','H','G','F','E','D']
```

The next cell involves using these lists to communicate to Pandas the correct order for the levels of the three categorical variables.

```python
diamonds.clarity = pd.Categorical(diamonds.clarity, clarity_levels)
diamonds.cut = pd.Categorical(diamonds.cut, cut_levels)
diamonds.color = pd.Categorical(diamonds.color, color_levels)
```

The next cell involves creating lists of named colors to serve as palettes to be used for visualizations later in the notebook.

```python
clarity_pal = ['tomato','moccasin','khaki','palegreen','skyblue','mediumpurple','orchid','lightpink']
color_pal = ['tomato','moccasin','khaki','palegreen','skyblue','mediumpurple','orchid']
cut_pal = ['tomato','moccasin','khaki','palegreen','skyblue']
```

## Part 5: Scatterplots of Price Against Carat
This section involves exploring the relationship between the price and carat attributes by generating scatter plots.

```python
plt.figure(figsize=[8,6])
plt.scatter(diamonds.loc[:,'carat'], diamonds.loc[:, 'price'], s = 20, alpha = 0.4, color = 'skyblue')
plt.xlabel('Carat Size')
plt.ylabel('Price (in $)')
plt.title('Relationship between Diamond Price and Carat Size')
plt.show()
```

The next cell involves reproducing the plot from above, but this time coloring the points according to their clarity.

```python
plt.figure(figsize=[8,6])

for i in range(0, len(clarity_levels)):
    sel = (diamonds['clarity'] == clarity_levels[i])
    plt.scatter(diamonds.loc[sel,'carat'], diamonds.loc[sel, 'price'], s = 20, alpha = 0.4, 
                color = clarity_pal[i], label = clarity_levels[i])
    
plt.xlabel('Carat Size')
plt.ylabel('Price (in $)')
plt.title('Relationship between Diamond Price and Carat Size')
plt.legend()
plt.show()
```

The next cell involves separating the points for each of the eight clarity levels into its own subplot.

```python
plt.figure(figsize=[12,6])

for i in range(0, len(clarity_levels)):
    sel = (diamonds['clarity'] == clarity_levels[i])
    plt.subplot(2,4,i+1)
    plt.scatter(diamonds.loc[sel,'carat'], diamonds.loc[sel, 'price'], s = 20, alpha = 0.4, 
                color = clarity_pal[i])
    
    plt.xlim([0,4])
    plt.ylim([0,20000])
    plt.xlabel('Carat Size')
    plt.ylabel('Price (in $)')
    plt.title(clarity_levels[i].title())

plt.tight_layout()
plt.show()
```

## Part 6: Applying Logarithmic Transformations
This section will start by adding two new columns to **diamonds** in order to store the transformed variables

```python
diamonds.loc[:,'ln_carat'] = np.log(diamonds.loc[:,'carat'])
diamonds.loc[:,'ln_price'] = np.log(diamonds.loc[:,'price'])
```

The next cell will use histograms to explore the distribution of the diamond prices and the log of the diamond prices.

```python
plt.figure(figsize = [12,4])
plt.subplot(1, 2, 1)
plt.hist(diamonds.loc[:,'price'], edgecolor = 'black', color = 'skyblue', bins = 20)
plt.xlabel('Price')
plt.ylabel('Count')
plt.title('Histogram of Prices')

plt.subplot(1, 2, 2)
plt.hist(diamonds.loc[:,'ln_price'], edgecolor = 'black', color = 'skyblue', bins = 20)
plt.xlabel('Natural Log of Price')
plt.ylabel('Count')
plt.title('Histogram of Log-Prices')

plt.show()
```

The next cell will use histograms to explore the distribution of the carat sizes and the log of the carat sizes.

```python
plt.figure(figsize = [12,4])
plt.subplot(1, 2, 1)
plt.hist(diamonds.loc[:,'carat'], edgecolor = 'black', color = 'tomato', bins = 20)
plt.xlabel('Carat Size')
plt.ylabel('Count')
plt.title('Histogram of Carat Sizes')

plt.subplot(1, 2, 2)
plt.hist(diamonds.loc[:,'ln_carat'], edgecolor = 'black', color = 'tomato', bins = 20)
plt.xlabel('Natural Log of Carat Size')
plt.ylabel('Count')
plt.title('Histogram of Log-Carat Sizes')

plt.show()
```

## Part 7: Scatterplot of Transformed Variables
The next cell will involve creating a scatterplot of ln_price against ln_carat, with points colored according to their clarity.

```python
plt.figure(figsize=[8,6])

for i in range(0, len(clarity_levels)):
    sel = (diamonds['clarity'] == clarity_levels[i])
    plt.scatter(diamonds.loc[sel,'ln_carat'], diamonds.loc[sel, 'ln_price'], s = 20, alpha = 0.4, 
                color = clarity_pal[i], label = clarity_levels[i])
    
plt.xlabel('Natural Log of Carat Size')
plt.ylabel('Natural Log Price (in $)')
plt.title('Relationship between Log-Price and Log-Carat Size')
plt.legend()
plt.show()
```

## Part 8: Grouping by Cut
This section involves creating a grouped DataFrame displaying the mean price and carat size for each cut level.

```python
gb_cut = pd.DataFrame(diamonds.loc[:,['cut','price','carat']].groupby('cut').mean())
gb_cut
```

The next cell will use bar charts to graphically display the information from the DataFrame above.

```python
plt.figure(figsize = [12,4])
plt.subplot(1, 2, 1)
plt.bar(gb_cut.index, gb_cut.loc[:,'price'], color = cut_pal, edgecolor = 'black')
plt.xlabel('Cut')
plt.ylabel('Mean Price')
plt.title('Mean Price by Cut')

plt.subplot(1, 2, 2)
plt.bar(gb_cut.index, gb_cut.loc[:,'carat'], color = cut_pal, edgecolor = 'black')
plt.xlabel('Cut')
plt.ylabel('Mean Carat Size')
plt.title('Mean Carat Size by Cut')

plt.show()
```
