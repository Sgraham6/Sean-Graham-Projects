```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

## Part 1: Iris Dataset

```python
iris.head(8)
```

```python
iris.describe()
```

```python
iris_means = iris.loc[:,['species', 'sepal_length', 'sepal_width', 'petal_length', 
                         'petal_width']].groupby('species').mean()
iris_means
```

```python
iris_colors = ['khaki', 'palegreen', 'skyblue']
iris_columns = ['Sepal Length', 'Sepal Width', 'Petal Length', 'Petal Width']

plt.figure(figsize = [8,6])
for i in range(0, len(iris_columns)):
    plt.subplot(2,2,i+1)
    plt.bar(iris_means.index, iris_means.iloc[:,i], color = iris_colors, edgecolor = 'black', label = iris_means.index)
    plt.title(iris_columns[i])
    
plt.tight_layout()
plt.plot()
```

```python
total_hr = pd.DataFrame(batting.loc[:,['teamID','yearID','HR']].groupby(['teamID','yearID'], as_index = False).sum())
total_hr = total_hr.sort_values(by='HR', ascending = False)
total_hr.head(10)
```

```python
avg_hr = pd.DataFrame(total_hr.loc[:,['teamID','HR']].groupby('teamID').mean())
avg_hr = avg_hr.sort_values(by='HR', ascending = False)
avg_hr.head(10)
```

## Part 7: Batting Data – Batting Averages

```python
batting_avg = pd.DataFrame(batting.loc[:,['teamID','yearID','H','AB']].groupby(['teamID','yearID']).sum())
batting_avg.loc[:,'BA'] = batting_avg.loc[:,'H'] / batting_avg.loc[:,'AB']
batting_avg = batting_avg.sort_values(by='BA', ascending = False)
batting_avg.head(10)
```

## Part 8: Batting Data – Cardinals vs. Cubs

```python
sel1 = (batting.loc[:,'teamID'] == 'SLN') & (batting.loc[:,'yearID'] >= 1900)
stl_batting = batting.loc[sel1,:]
stl_batting = pd.DataFrame(stl_batting.loc[:,['yearID','H','AB','HR']].groupby('yearID').sum())
stl_batting.loc[:,'BA'] = stl_batting.loc[:,'H'] / stl_batting.loc[:,'AB']

sel2 = (batting.loc[:,'teamID'] == 'CHN') & (batting.loc[:,'yearID'] >= 1900)
chi_batting = batting.loc[sel2,:]
chi_batting = pd.DataFrame(chi_batting.loc[:,['yearID','H','AB','HR']].groupby('yearID').sum())
chi_batting.loc[:,'BA'] = chi_batting.loc[:,'H'] / chi_batting.loc[:,'AB']
```

```python
plt.figure(figsize = [12,4])

plt.subplot(1,2,1)
plt.plot(stl_batting.index, stl_batting.loc[:,'BA'], color = 'skyblue', label = 'Cardinals')
plt.plot(chi_batting.index, chi_batting.loc[:,'BA'], color = 'palegreen', label = 'Cubs')
plt.legend()
plt.xlabel('Year')
plt.ylabel('Batting Average')
plt.title('Batting Average By Year')

plt.subplot(1,2,2)
plt.plot(stl_batting.index, stl_batting.loc[:,'HR'], color = 'skyblue', label = 'Cardinals')
plt.plot(chi_batting.index, chi_batting.loc[:,'HR'], color = 'palegreen', label = 'Cubs')
plt.legend()
plt.xlabel('Year')
plt.ylabel('Home Runs')
plt.title('Home Runs By Year')

plt.tight_layout()
plt.show()
```

```python
comp1 = np.mean(stl_batting.loc[:,'BA'] > chi_batting.loc[:,'BA'])
print(round(comp1, 4))
```
