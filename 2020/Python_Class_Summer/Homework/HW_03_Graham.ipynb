### Part 1: Dot Product

Write a function called `dot_product()` that takes two parameters called `x` and `y`. The function should multiply each element of `x` with the corresponding element of `y`, sum these products, and then return the sum.

For instance, if `a` and `b` are as follows:

    a = [3, 4, 2]
    b = [5, 2, 3]

then `dotProduct(a,b)` should return `29` (since `15 + 8 + 6 = 29`). 

```python
def dot_product(x, y):
    total = 0
    for i in range (0, len(x)):
        total += x[i] * y[i]
    return total
```

### Part 2: Amortization

Define a function called `count_payments()`. The function should take three parameters: `amount`, `rate`, and `pmt`. 
* `amount` is the size of a loan that will be repaid with monthly payments. 
* `rate` is the monthly interest rate. 
* `payment` is the size of the monthly payment. 

Each month, the amount still owed will be multiplied by `(1 + rate)`, and then `pmt` will be subtracted from the amount owed. This process will repeat again every month, until the amount owed is less than or equal to zero. Use a loop to simulate this process.

The function should return the number of monthly payments that need to be made in order to repay the loan. 


```python
def count_payments(amount, rate, pmt):
    count = 0
    while amount >= 0:
        amount = amount*(1 + rate)-pmt
        count += 1
    return count
```

### Part 3: Minimum

Write a function called `minimum()`. The function should take a single parameter, called `x`, and should the return the smallest element of `x`.

For example, if `b = [7.5, 8.7, 4.2, 9.1]`, then `minimum(b)` should return 4.2. 

You may use the `len` function in this problem.

```python
def minimum(x):
    min_val = x[0]
    for i in range (0, len(x)):
        if x[i] < min_val:
            min_val = x[i] 
    return min_val
```

### Part 6: Find Unique Elements

Write a function called `unique` with a single parameter `x`. The function should return a new list, containing a single occurence of every value that appears in `x`. The list returned should be sorted in ascending order. 

For example, if `mylist = [4, 6, 3, 3, 9, 6, 4]`, then `unique(mylist)` should return `[3, 4, 6, 9]`.

**Hint:** I would recommend using one or both of the following comparison operators: `in` and `not in`. This operator checks to see if a particular value appears within a list. For example, if `mylist` is as defined above, then:
* `4 in mylist` would evaluate as `True`, and  `4 not in mylist` would evaluate as `False`. 
* `2 in mylist` would evaluate as `False`, and  `2 not in mylist` would evaluate as `True`.

You are allowed to use the `append` and `sort` list methods, but no other predefined functions or methods. 

```python
def unique(x): 
    unique_list = [] 
    for i in x: 
        if i not in unique_list: 
            unique_list.append(i)
    unique_list.sort()
    return(unique_list)
```

### Part 7: Find Frequency Distribution

Write a function called `freq_dist` with a single parameter `x`. This function should return two lists. You can name these whatever you like, but for the sake of discussion, lets call them `values` and `counts`. 
* The list `values` should be a list of unique values appearing in `x`. You can use the function you wrote in Part 6 to create this list. 
* The list `counts` should be a list of integers. Each integer should be a count of the number of times a particular value in `values` appears in `x`. 

For example: If `mylist = [10, 10, 20, 10, 20, 30, 20, 20, 30, 40]`, then `freq_dist(mylist)` should return the following tuple:
    `([10, 20, 30, 40], [3, 4, 2, 1])`. 

**Hint:** In addition to the `unique` function that you defined previously, you may also use the `append` and `count` list methods. The `count` method can be used to count the number of times a particular value appears within a list. You can read more about that method [`here`](https://www.tutorialspoint.com/python/list_count.htm).

```python
def freq_dist(x):
    values = unique(x)
    counts = []
    for i in values: 
        counts.append(x.count(i))
    return values, counts
```

### Part 8: Weighted Average

Write a function called `weighted_mean` that accepts two parameters `x` and `w`. The parameter `x` is expected to be a list of numerical values. The parameter `w` is expected to be either have a value of `None`, or to be a list of numerical values. The default value of `w` should be `None`. If `w` is `None`, then the function should return the mean of the elements in the list. If `w` is a list, then the function should return the weighted mean of `x` using the elements of `w` as the weights. That is to say, the function should return the following:

$$\frac{w_1 x_1 + w_2 x_2 + ... + w_n x_n }{w_1 + w_2 + ... + w_n}$$

As an example, if:

    list1 = [200, 400, 100]
    list2 = [3, 2, 5]
    
Then `weighted_mean(list1, list2)` should return:

$$\frac{200 (3) + 400 (2) + 100(5)}{3 + 2 + 5} = \large \frac{1900}{10} = 190$$ 

You may use the `sum` and `len` functions in this problem. 

```python
def weighted_mean(x, w = None):
    if w == None:
        answer = sum(x) / len(x)
    else:   
        numerator = 0
        for i in range (0, len(x)):
            numerator += x[i] * w[i]
            denominator = sum(w)
            answer = numerator / denominator
    return answer
```

Two lists are provided in the cell below. Calculated the mean of `values`, as well as the weighted mean of `values` obtained by using the weights stored in `weights`. Print the results in the following format:


    Standard Mean: xxxx
    Weighted Mean: xxxx

```python
values = [4, 7, 3, 5, 2, 6, 8, 2, 4, 6]
weights = [2, 1, 4, 1, 2, 2, 1, 4, 3, 2]

print("Standard Mean: " + str(weighted_mean(values)))
print("Weighted Mean: " + str(weighted_mean(values, weights)))
```
