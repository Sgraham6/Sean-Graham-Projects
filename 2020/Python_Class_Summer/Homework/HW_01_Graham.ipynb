### Part 4: Probability

Let `P(A or B)` denote the probability of at least one of the two events occurring, and let `P(A and B)` denote the probablity of both events occurring. 

Two events are said to be **independent** if the knowledge that one of the events has occurred has no effect on the probability that the other event has occurred.

Probability theory tells us the following:

* `P(A or B)` and `P(A and B)` are related by the following formula: **`P(A or B) = P(A) + P(B) - P(A and B)`.**
*  If `A` and `B` are independent, then **`P(A and B) = P(A) * P(B)`.** 

-----

# Part 5: Bayes' Rule

Assume that you are studying the effectiveness of a test used to determine whether or not someone has developed a specific viral infection. Suppose you know the following information about the test:

* The probability of someone testing positive for the infection, assuming that they actually have it is $P(\textrm{Test Pos} ~|~ \textrm{Infected}) = 0.90$.
* The probability of someone testing positive for the infection, assuming that they do not have it is $P(\textrm{Test Pos} ~|~ \textrm{Healthy}) = 0.10$.
* If a person is selected at random from the population, the probability that they will have the infection is $P(\textrm{Infected}) = 0.14$. 

In the cell below, create the following variables to store the probabilities mentioned above: `prob_pos_if_infected`, `prob_pos_if_healthy`, and `prob_infected`.


```python
prob_pos_if_infected = 0.90
prob_pos_if_healthy = 0.10
prob_infected = 0.14
print(prob_pos_if_infected, prob_pos_if_healthy, prob_infected)
```

Probability theory tells us that:

* If a person is selected at random from the population, the probability that they will NOT have the infection is given by $P(\textrm{Healthy}) = 1 - P(\textrm{Infected})$.
* The probability that a randomly selected person will have the infection, and will also test positive is given by $P(\textrm{Test Pos and Infected}) = P(\textrm{Test Pos} ~|~ \textrm{Infected}) \cdot P(\textrm{Infected})$.
* The probability that a randomly selected person will NOT have the infection, but will test positive is given by $P(\textrm{Test Pos and Healthy}) = P(\textrm{Test Pos} ~|~ \textrm{Healthy}) \cdot P(\textrm{Healthy})$.
* If a person is selected at random from the population, the probability they they will test positive for the infection is given by $P(\textrm{Test Pos}) = P(\textrm{Test Pos and Infected}) + P(\textrm{Test Pos and Healthy})$

Use the cell below to calculate each of these four probabilities, storing the results in variables named `prob_healthy`, `prob_pos_and_infected`, `prob_pos_and_healthy`, and `prob_pos`. Use the variables defined in the previous cell in your calculations.


```python
prob_healthy = 1 - prob_infected
prob_pos_and_infected = prob_pos_if_infected * prob_infected
prob_pos_and_healthy = prob_pos_if_healthy * prob_healthy
prob_pos = prob_pos_and_infected + prob_pos_and_healthy
print(prob_healthy, prob_pos_and_infected, prob_pos_and_healthy, prob_pos)

#since my first print line produced unnecessary digits, I added a line where the values are properly rounded
print((round(prob_healthy, 2)),(round(prob_pos_and_infected, 3)),(round(prob_pos_and_healthy, 3)),(round(prob_pos, 3)))
```

Now, assume that a person has tested positive for the virus. The probability that they actually have the virus is given by: $\Large\frac{P(\textrm{Test Pos and Infected})}{P(\textrm{Test Pos})}$.

Calculate this probability, storing the result in a variable called `prob_infected_if_pos`. 

Print a message reporting the result of this final calculation. The message should be in the format shown below, with the `xxxx` symbols replaced with the appropriate values. **Round the probability displayed to four decimal places.**

```python
prob_infected_if_pos = prob_pos_and_infected / prob_pos
print("Probability of being infected, given a positive test result: " + str(round(prob_infected_if_pos, 4)))
```
