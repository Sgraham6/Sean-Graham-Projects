### Part 3: Scoring a Model with Sum of Squared Errors

A biologist is studying the relationship between the length and weight of yellow-bellied marmots. Based on prior work, she believes that the length and weight of adult marmots can be approximately modeled by a relationship of the following form:

$$weight = 0.6 * length - 5.2$$

To test this model, the biologist collects a sample of two adult marmots. The lengths and weights of the marmots are provided in the cell below. Run this cell as-is. 


```python
length = [22.7, 22.4, 25.8, 21.3, 20.1, 22.1, 21.1, 25.3, 26.9, 26.9,
          23.0, 23.8, 26.2, 20.4, 23.0, 21.9, 23.5, 27.8, 25.3, 25.9]

weight = [9.2, 8.8, 10.7,  8.3, 6.2, 8.6, 7.2, 11.2, 10.5, 11.3, 
          9.6, 9.9, 10.9,  5.9, 9.5, 9.1, 9.7, 11.6, 10.2, 10.5]
```

In the cell below, create a list called `pred_weight` that contains the predicted weight for each marmot, according to the biologist's model. Print `pred_weight`. 

**Hint:** `pred_weight` should start as an empty list. Loop over the elements of `length`. Each iteration of the loop should calculate a new predicted weight and append it to the list `pred_weight`. 

```python
pred_weight = []

for i in range (0, len(length)):
    pred = 0.6 * length[i] - 5.2
    pred_weight.append(pred)
    
print(pred_weight)
```

The biologist scores her model using the sum of squared errors (SSE) metric. This metric is calculated as follows:
1. For each observation, calculate the difference between the true weight and the predicted weight. 
2. Square this difference. 
3. Sum the squared differences over all observations. 

Calculate the model's `SSE` score for this sample. There are multiple ways to perform this task, but for the sake of efficiency, I would like for you to do it **without creating any new lists** and using **only one loop**. Print the value of `SSE` rounded to four decimal places. 

**Hint**: Loop over the elements of weight. For each element, calculate the error, square it, and then add it to a running total. 

```python
total = 0
for i in range (0, len(weight)):
    total += (weight[i] - pred_weight[i])**2
    
print(round(total, 4))
```

### Part 4: Calculating Accuracy

A statistical model is developed for the purposes of detecting the presence of a disease in patients based on the results of blood work and other medical information. To test the performance of the model, it is applied to 20 individuals for which the correct diagnosis is already known. 

Two lists are provided in the cell below. The list `true_class` stores the correct diagnoses for the 20 individuals. The list `pred_class` contains the diagnoses predicted by the model. In both cases, an entry of `'P'` indicates a positive diagnosis (indicating the individual does have the disease) while an entry of `'N'` indicates a negative result (indicating the individual does not have the disease). The entries in the two arrays are paired so that for any index `i`, the results stored in `true_class[i]` and `pred_class[i]` refer to the same individual. 

```python
true_class = ['P', 'P', 'N', 'N', 'P', 'N', 'N', 'N', 'P', 'N',
              'N', 'N', 'N', 'P', 'P', 'N', 'N', 'N', 'N', 'N']

pred_class = ['N', 'P', 'N', 'P', 'P', 'N', 'P', 'N', 'P', 'N',
              'N', 'N', 'P', 'P', 'P', 'N', 'N', 'N', 'P', 'N']
```

The **accuracy** of the model on the supplied data is equal to the proportion of predictions that were correct. Add code to the cell below to calculate the model's accuracy on this group of individuals. First count the number of correct predictions, and then divide by the total number of individuals. You will need to use a single loop, as well as conditional statements. Print the results in the following format:

    Model Accuracy: XXXX
    
**Provide your answer as a decimal, rounded to 4 decimal places.**

```python
correct = 0
incorrect = 0

for i in range (0, len(true_class)):
    if true_class[i] == 'P' and pred_class[i] == 'P':
        correct += 1
    elif true_class[i] == 'N' and pred_class[i] == 'N':
        correct += 1
    else:
        incorrect += 1
        
accuracy = correct / (correct + incorrect)
print("Model Accuracy: " + str(round(accuracy, 4)))
```

### Part 7:  Calculating Risk Rating

The cell below creates two large lists with the same (unknown) size: `amount` and `rating`. Please run this cell as is.

```python
import numpy as np
np.random.seed(1)
n = np.random.choice(range(120,340))
amount = list(np.round(np.random.normal(2000, 600, n),0))
rating = list(np.random.choice(['L', 'M', 'H'],n))
del n
```

A small lending agency called LendCo makes personal and small business loans. The size of the loans for all of LendCo's clients are stored in the list `amount`. LendCo also assigns a risk rating of low, medium, or high for each of its clients. These ratings are stored in the list `rating` as strings `L`, `M`, or `H`. 

At any given time, LendCo calculates its current overall "risk rating" as follows:
* The amount of any low risk loan is multiplied by 1.
* The amount of any medium risk loan is multiplied by 1.4.
* The amount of any high risk loan is multiplied by 2. 
* The results from above are summed together for all of the loans. 

Calculate LendCo's current risk rating, rounded to 1 decimal place. Print the result in the following format:

    LendCo's risk rating: xxxxxx.x

**To recieve full credit, your solution should involve only one loop, and should not create any new lists.**

```python
risk_rating = 0

for i in range (0, len(amount)):
    if rating[i] == 'L':
        adjusted_rating = amount[i]
        risk_rating += adjusted_rating
    elif rating[i] == 'M':
        adjusted_rating = amount[i] * 1.4
        risk_rating += adjusted_rating
    else:
        adjusted_rating = amount[i] * 2
        risk_rating += adjusted_rating

print("LendCo's risk rating: " + str(round(risk_rating, 1)))
```

```python
A = [ [13, 43, 28, 22, 41], [17, 39, 46, 16, 21], [41, 34, 31, 25, 14] ]
B = [ [35, 29, 43, 21, 31], [48, 26, 19, 17, 23], [32, 34, 24, 16, 27] ]
```

```python
AplusB = []

for i in range (0, 3):
    sub_AplusB = []
    for j in range (0, 5):
        sub_AplusB.append(A[i][j]+B[i][j])
    AplusB.append(sub_AplusB)
```
